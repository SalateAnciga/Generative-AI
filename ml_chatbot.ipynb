# Rule Based Moderation

```python
def moderate_question(user_question):
    
    blocked_words = [
        "violence",
        "illegal",
        "hack"
    ]
    
    for word in blocked_words:
        if word.lower() in user_question.lower():
            return False
    
    return True
```

```python
print(moderate_question("what is  machine learning"))
```

```python
user_question = input("Ask a question: ")

if moderate_question(user_question):
    print("Question is allowed.")
else:
    print("Sorry, I cannot process this question.")
```

# Topic Moderation

```python
system_prompt = f"""
You are a Machine Learning Assistant.

Your ONLY purpose is to answer questions related to Machine Learning,
Artificial Intelligence, Deep Learning, NLP, and Data Science.

IMPORTANT RULE:
If the user's question is NOT related to Machine Learning or the allowed topics,
DO NOT answer the question and DO NOT provide any additional information.

For any unrelated question, reply with EXACTLY this sentence:

"I am a Machine Learning Assistant. Please ask a question related to Machine Learning."

Do not provide examples, suggestions, explanations, recipes, or any other information
for unrelated questions.

Machine Learning information:
{ML_INFO}
"""
```

# Generate the Answer

```python
user_question = input("Ask a Machine Learning question: ")

response = ollama.chat(
    model="llama3.2",options={"temperature":0, "num_predict":500},
    messages=[
        {
            "role": "system",
            "content": system_prompt
        },
        {
            "role": "user",
            "content": user_question
        }
    ]
)
answer=response["message"]["content"]
print("\nML Assistant:")
print(answer)
```

# Evaluation

```python
evaluation_prompt = """
You are an evaluator for a Machine Learning Assistant chatbot.

Evaluate the chatbot answer using this rubric.

1. Correctness: Is the answer technically correct? Give 1-5.
2. Relevance: Does it directly answer the user's question? Give 1-5.
3. Clarity: Is it easy for a beginner to understand? Give 1-5.
4. Completeness: Does it contain the important information? Give 1-5.

Give:
- Score for each criterion
- Total score out of 20
- Short explanation
"""
```

```python
evaluation_response = ollama.chat(
    model="llama3.2",
    messages=[
        {
            "role": "system",
            "content": evaluation_prompt
        },
        {
            "role": "user",
            "content": f"""
User Question:
{user_question}

Chatbot Answer:
{answer}
"""
        }
    ]
)

evaluation = evaluation_response["message"]["content"]

print("\nEvaluation:")
print(evaluation)
```
